```
### Eksempler:

#### `np.ones((4, 3, 2))`:
```
4 layers (4 "bøker")
3 rows per layer (3 linjer per side)
2 columns per row (2 tall per linje)

Total elementer: 4 × 3 × 2 = 24 elementer
```

#### `np.zeros((4, 3, 2))`:
```
Samme structure, men fylt med 0 i stedet for 1
```

#### `np.random.random((4, 3, 2))`:
```
Samme structure, men fylt med tilfeldige tall (0 til 1)

📝 Kodeeksempel:

````python
import numpy as np

# Lag en 3D array med ones
arr= np.ones((4, 3, 2))

print(f"shape: {arr.shape}")
print(f"Dimensions: {arr.ndim}")
print(f"Size: {arr.size}")

# Print hele arrayen
print(arr)
````

🎯 Indexing i 3D arrays:

````python
arr_3d = np.array([
    [[1, 2], [3, 4]],    # Layer 0
    [[5, 6], [7, 8]]     # Layer 1
])

# Syntaks: arr[layer, row, column]

# Hent ett element:
print(arr_3d[0, 0, 0])  # 1 (layer 0, rad 0, kolonne 0)
print(arr_3d[0, 0, 1])  # 2 (layer 0, rad 0, kolonne 1)
print(arr_3d[1, 1, 1])  # 8 (layer 1, rad 1, kolonne 1)

# Hent en hel layer:
print(arr_3d[0])
# [[1 2]
#  [3 4]]

# Hent en rad fra en layer:
print(arr_3d[0, 0])     # [1 2] (layer 0, rad 0)

# Hent en kolonne fra alle layers:
print("en kilonne fra alle layers:", arr_3d[:, 0, 0])
````

````python
# Et 100x100 RGB-bilde er en 3D array:
image = np.random.randint(0, 255, size=(100, 100, 3))
#                                       ↑    ↑   ↑
#                                       │    │   └─ 3 fargekanaler (R, G, B)
#                                       │    └───── 100 rader (høyde)
#                                       └────────── 100 kolonner (bredde)

# Layer 0 = Rød kanal
# Layer 1 = Grønn kanal  
# Layer 2 = Blå kanal

print(image)
````
