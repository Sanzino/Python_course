### Creating tuples
In this section, I create tuples using parentheses and verify their data type.


```python
# Code example / demonstration
numbers = (1, 2, 3, 4, )
languages = ("Python", "Java", "C++")

print (numbers)
print (languages)
print(type(numbers))
```

### Indexing tuples
In this section, I access elements in a tuple using indexing.
Indexing rules are the same as for lists and strings.


```python
print(languages[0])  #first element
print(languages[1]) #second element
print(languages[-1]) #last element
```

### Slicing tuples
In this section, I extract parts of a tuple using slicing.
The start index is included and the end index is excluded.


```python
print(languages[0:2])  #slicing    
print(languages[1:])    #slicing from second to end
```

### Tuple immutability
In this section, I demonstrate that tuples cannot be modified after creation.


```python
# This will fail if uncommented
# languages[0] = "Go"

print("Tuples are immutable")
```

### Tuples vs Lists
In this section, I compare tuples and lists.


```python
my_list = ["Python", "Java"]
my_tuple = ("Python", "Java")

my_list[0] = "Go"
print("List:", my_list)

# This would fail:
# my_tuple[0] = "Go"
print("Tuple:", my_tuple)
```

## 📝 Key Takeaways
- Tuples store multiple values in a single variable
- Tuples are ordered and support indexing and slicing
- Tuples are immutable and cannot be changed after vreation
- Tuples are useful for fixed or constant data
- Lists should be used when data needs to be modified
