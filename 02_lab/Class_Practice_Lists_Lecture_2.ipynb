#### Creating lists

In this section, I create a list to store multiple values in a single variable.
A list is written using square brackets `[]` and can contain multiple elements.

I also use `type()` to confirm that the variable is a list.


```python
items = ["apple", "banana", "cherry"]
print(items)
print(type(items))
```


### Indexing lists

In this section, I access individual elements in a list using indexing.
Indexing starts at 0, meaning the first element has index 0.

Negative indexing is also supported, where `-1` refers to the last element in the list.


```python
print(items[0])   #First element
print(items[1])   #Second element
print(items[-1])  # also last element

```

### Slicing lists

In this section, I extract parts of a list using slicing.
Slicing uses the format `start:end`, where the start index is included and the end index is excluded.

Slicing makes it easy to work with subsets of a list.


```python
print(items[0:2]) 
print(items[1:])   # Accessing the first item
```

### List mutability

In this section, I modify an element in the list.
Lists are mutable, meaning their elements can be changed after the list is created.

This is an important difference between lists and strings.


```python
items[1] = "pear"
print(items)

```

### List methods

In this section, I use common list methods to modify the list.

- `append()` adds a new element to the end of the list  
- `remove()` removes a specific element by value  
- `pop()` removes and returns the last element in the list  

These methods change the original list.


```python
items.append("grape")
print(items)

items.remove("apple")
print(items)

last_item = items.pop()
print(last_item)
print(items)
```

## 📝 Key Take aways
## 📝 Key Takeaways

- A list is a data structure used to store multiple values in a single variable
- Lists are ordered and support indexing and slicing
- Indexing starts at 0 and negative indices count from the end
- Lists are mutable, meaning elements can be changed after creation
- Common list methods such as `append()`, `remove()`, and `pop()` modify the list
- Lists are commonly used for storing and processing collections of data
