# 2. String indexing
Each character in a string has an index starting at 0.
Negtiv indices count from the end

```python
word = 'Python'

word[0]  # 'P'
word[1]  # 'y'
word[-1] # 'n'
word[-2] # 'o'

print(word[0])
print(word[1])
print(word[-1])
print(word[-2])

print(type(word[1]))


```

# 3. String slicing 

Sclicing extracts parts of a string.
Start indec is includedm end index is excluded.

| Index | 0 | 1 | 2 | 3 | 4 | 5 |
| ----- | - | - | - | - | - | - |
| Char  | P | y | t | h | o | n |

word[0:2]  # characters at index 0 and 1
word[2:6]  # characters at index 2,3,4,5

⚠️ Common mistakes (EXAM TRAPS)

❌ Expecting end index to be included
word[0:3]  # 'Pyt', NOT 'Pyth'

❌ Out-of-range slicing (this is actually safe)
word[2:100]  # 'thon' (no error)


```python
text = "Cybersecurity"

text[0:5]      # 'Cyber'   
text[-5:]      # 'rity'
text[:8]       # 'Cybersec'
text[8:]       # 'urity'

print(text[0:5])
print(text[-5:])
print(text[:8])
print(text[8:])
```

# 4. String Immutability (Exam trap)

What does immutability mean?
it means that a string cannot be changed after it is created.

⚠️ Common misunderstanding

This does NOT mean variables are immutable:

word = "Python"
word = "Java"

once a string is assigned to a variable, its individual chracters cannot be modified.

Code example 1 will fail.
Code example 2 is correct approach.




```python
word = "Python"
word[0] = "J"
```

```python
word = "Python"
new_word = "j" + word[1:]
new_word  # 'jython'
print(new_word)
```

🧠 Important mental model

You cannot change a string
You can only create a new string based on the old one


# 5. String Methods

What are string method?
string methods are built-in functions that oprerate on strings and retun new strings.

because strings are immutable, methods do not modify the original string.

🔑 Very important rule (EXAM)

String methods return new strings — they do NOT change the original string

Proof (Code cell)
print(text)


Output:

 hello python 

 ⚠️ Common mistakes (exam traps)

❌ Expecting methods to change the string
text.upper()
print(text)   # Still unchanged

✅ Correct usage

text = text.upper()
print(text)



```python
message = "   Cyber Security    "
print(message.strip())      # 'Cyber Security'
print(message.lower())      # ' cyber security '
print(message.replace(" ", "_"))  # 'CyberSecurity'
print(type(message.split()))  # <class 'list'>
```
