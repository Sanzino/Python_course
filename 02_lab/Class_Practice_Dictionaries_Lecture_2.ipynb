### Creating dictionaries
In this section, I create a dictionary and inspect its structure.


```python
student = {
    "name": "Shaffi",
    "age": 24,
    "program": "IT"
}   

print(student)
print(type(student))

```

### Accessing values
In this section, I access dictionary values using keys.


```python
print(student["name"])
print(student["age"])
```

### Modifying values
In this section, I modify an existing value in the dictionary.


```python
student["age"] = 23
print(student)
```

### Adding new items
In this section, I add a new key–value pair to the dictionary.


```python
student["grade"] = "A"
print(student)
```

### Removing items
In this section, I remove a key–value pair from the dictionary.


```python
student.pop("program")
print(student)
```

## 📝 Key Takeaways

- Dictionaries store data using key-value pairs 
- Values are accessed using keys, not indices
- Dictionaries are mutable and can be modified
- Key must be unique and immutable
- Dictionarie are useful for structured and labled data
