## 💻 Code Example / Demonstration

## Creating a greet function


```python
def greet():
    print("hello!")

greet()
```

### Creating a FUnction with parameters


```python
def greet_user(name):
    print("hello", name)

greet_user("Alice")
```

## Function with return value

```python
def add(a, b):
    return a + b    

result = add(3, 5)
print(result)  # prints 8
```

🧠 Concept Explanation (in my own words)

A loop allows a program to repeat a block of code multiple times. In pyton, a for loop is commonly used to iterate over a squence such as a list or a string

Instea of writing the same code repeatedly, a lopp runs the vode once for each item in the sequence.

Loops are essential when working with collection of data.



💻 Code Example / Demonstration

##Looping through a list

```python
items = ["apple", "banana", "cherry"]

for item in items:
    print(item)
    
```

## Looping through a string


```python
word = "Python"

for letter in word:
    print(letter)

```

## Using range()

```python
for i in range(5):
    print(i)
    
```

⚠️ Common Mistake (Exam Relevant)

```python
for i in 5:
    print(i)


## ❌ This fails because 5 is not iterable.
## Use range(5) instead.
```

🧠 Concept Explanation (in my own words)
Working with data means using loops, conditionals, and data structures together to analyze or transform information.

Instead of working with single values, programs often prcess colletions of data such as lists or dictionaries.

By combining "for loops" with "if" statements, we can filter, count, or modify data efficently

#💻 Code Example / Demonstration

#Example 1 filtering data from a list

```python


numbers = [1, 4, 7, 10, 13]

for number in numbers:
    if number > 5:
        print(number)  # prints 7, 10, 13
```

#Example 2: Processing dictionary data


```python
student_scores = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
}

for name, score in student_scores.items():
    if score >= 80:
        print(name)  # prints Alice and Bob
```

🧠 Common Data Patterns (recognize these)

Filtering → selecting some values

Counting → keeping a counter

Updating → modifying values

Searching → finding specific data
