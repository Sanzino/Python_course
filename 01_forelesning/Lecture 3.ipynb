# Del 1: Python Modules - Grunnleggende Import

```python
# EKSEMPEL 1: Import av math module
# math er et built in module i Python som inneholder matematiske funksjoner

import math

# Når vi bruker "import math", må vi perfikse alle funksjoner med 'math.' 
resultat = math.sqrt(16)  # kvadratrot av 16
print(f"kvadratrot av 16 er: {resultat}")

# Andre nyttige math-funksjoner:
print(f"pi er: {math.pi}")  # Verdien av pi
print(f"Cosinus av 0: {math.cos(0)}")  # Cosinus av 0 grader
print(f"Logartime av 8 (base e): {math.log(8)}")  # Naturlig logaritme av 8
print(f"Logartime av 8 (base 2): {math.log2(8)}")  # Logaritme av 8 med base 2

```

```python
# EKSEMPEL 2: import med alias
# Vi kan gi modulene kortere navn for å spare skriving

import math as m

# Nå kan vi bruke 'm' i stedet for 'math'
print(f"Kvadratrot av 25: {m.sqrt(25)}")  # kvadratrot av 25
```

```python
# EKSEMPEL 3: Import av spesefikke funksjoner 
# Vi kan importere bare de funkjsonene vi trenger 

from math import sqrt, pi

# Nå kan vi bruke sqrt og pi direkte uten 'math.' prefix
print(f"Kvadratrot av 9: {sqrt(9)}")  # kvadratrot av 9
print(f"Verdien av pi: {pi}")  # Verdien av pi

# MERK: Dette fungerer ikke lenger fordi vi ikke importerte cos:
# print(f"Cosinus av 0: {cos(0)}")  # Dette vil gi en feil fordi cos ikke er importert

```

```python
# Eksempel 4: Lage en egen funkjson som bruker math 
# Fra forelesning 2 vet vi at vi kan lage egne funksjoner. Vi kan også bruke funksjoner fra math i våre egne funksjoner.

import math

def hyp(a, b):
    """
    Beregn hypotenusen i en rettvinklet trekant.
    
    Parameters:
    a (float): lengden av den ene kateten
    b (float): lengden av den andre kateten

    Returns:
    float: lengden av hypotenuse (c)

    Pythagoras' teorem sier at c² = a² + b²
    Derfor: c = √(a² + b²)
    """
    res = math.sqrt(a**2 + b**2) # Beregn kvadratrot av (a² + b²)
    return res

# Test funksjonen
print(f"Hypotenusen til en trekant med kateter 3 og 4 er: {hyp(3, 4)}")  # Skal returnere 5
print(f"Hypotenusen til en trekant med kateter 5 og 12 er: {hyp(5, 12)}")  # Skal returnere 13  
```

# Del 2: Algebra Grunnlag - Scalars, Vectors, Matrices, Tensors

```python
# SCALAR (0-dimensjonal)
# Et enkelt tall

# Fjernet import som skapte navnkonflikt med variabelen 'matric'

scalar = 5
print(f"scalar: {scalar}")
print(f"type: {type(scalar)}")
print()

# VECTOR (1-dimensjonal) - en liste med tall
# I Pyhton kan vi representere dette som en list

vector = [1, 2, 3, 4, 5]
print(f"Vector: {vector}")
print(f"type: {type(vector)}")
print()

# MATRIX (2-dimensjonal) - en tabell med tall (rader og kolonner)
# I Pyhton kan vi representere dette som en liste av lister

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(f"Matrix:")
for row in matrix:
    print(row)
print(f"Antall rader: {len(matrix)}")
print(f"Antall kolonner: {len(matrix[0])}")
print()

# TENSOR (3-dimensjonal eller høyere) - en struktur med flere dimensjoner
# For eksempel: 2 matriser stablet oppå hverandre

tensor = [
    [[1, 2], [3, 4]],  # Første matrise
    [[5, 6], [7, 8]]   # Andre matrise
]
print(f"Tensor (3D):")
for i, layer in enumerate(tensor):
    print(f"Layer {i+1}:") 
    for row in layer:
        print(f"  {row}")
```

# Del 3: Data Science Packages - NumPy

```python
# NUMPY - Numerical Python
# Brukes for: Arrays, matrices, matematiske operasjoner
# Standard import: import numpy as np

import numpy as np

# Lag en NumPy array (bedre enn Python lister for tall)

np_array = np.array([1, 2, 3, 4, 5])
print(f"NumPy array: {np_array}")
print(f"Type: {type(np_array)}")
print()

# Fordeler med NumPy:
# 1. Raskere beregninger
# 2. Vectorized operations (operasjoner på hele arrays samtidig)

print("Vectorized operations:")
print(f"Alle tall *2: {np_array * 2}")  # Multipliserer alle elementer i arrayen med 2
print(f"Alle tall + 10: {np_array + 10}")  # Legger til 10 til alle elementer i arrayen
print(f"Alle tall ** 2: {np_array ** 2}")  # Kvadrerer alle elementer i arrayen


```

# Del 4: Data Science Packages - Pandas


```python
# PANDAS - Panel Data
# Brukes for: Tabulære data (som Exel-tabeller)
# Standard import: import pandas as pd

import pandas as pd

# Lag en enkel DataFrame (tabell) 

data = {
    'Navn': ['Alice', 'Bob', 'Charlie'],
    'Alder': [25, 30, 35],
    'By': ['Oslo', 'Bergen', 'Trondheim']

}

df = pd.DataFrame(data)
print("Pandas DataFrame:")
print(df)
print()
print(f"Type: {type(df)}")
print()

# Noen grunnleggende operasjoner: 
print(f"Antall rader og kolonner: {df.shape}")
print(f"kolonnenavn {df.columns.tolist()}")
print()
print("Første rad:")
print(df.head(1))  # Vis første rad
```

# Del 5: Data Science Packages - Matplotlib

```python
# MATPLOTLIB - Plotting/Visualisering
# Brukes for: Lage grafer og plots
# Standard import: import matplotlib.pyplot as plt

import matplotlib.pyplot as plt
import numpy as np

# Enkelt eksempel: Line plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, marker='o', color='blue', linewidth=2) # Lag en linje-plot med prikker  
plt.title("Enkel Line Plot")
plt.xlabel("X-akse")
plt.ylabel("Y-akse")
plt.grid(True)
plt.show()  # Vis plottet

print("Grafen vises over")


# Forklaring:

# plt.plot() - Lager linje-grafen
# marker='o' - Legger til prikker på datapunkter
# color='blue' - Setter fargen til blå
# plt.grid(True) - Viser rutenett
# plt.show() - Viser grafen
```

# Del 6: Sammenligning - Python Lists vs NumPy Arrays


```python
import numpy as np

# PYTHON LIST = Fleksibel, men treg for matematikk
python_list = [1, 2, 3, 4, 5]

# For å doble alle tall i en liste, må vi bruke loop eller list comprehension
doubled_list = [x * 2 for x in python_list]
print(f"Python list doblet: {doubled_list}")
print()

# NUMPY ARRAY = Raskere og mer effektivt fordi den er spesilasert for tall
numpy_array = np.array([1, 2, 3, 4, 5])

# Med NumPy kan vi bare multiplisere direkte (vectorized operations)
doubled_array = numpy_array * 2
print(f"NumPy array doblet: {doubled_array}")
print()

# Andre fordeler med NumPy - innebygde funkjsoner:
print("NumPy innebygde funksjoner:")
print(f"Sum av array: {numpy_array.sum()}")  # Sum av alle elementer
print(f"Gjennomsnitt: {numpy_array.mean()}")  # Gjennomsnitt av elementene
print(f"Standardavvik: {numpy_array.std()}")
print(f"Minimum: {numpy_array.min()}")
print(f"Maximum: {numpy_array.max()}")
print(f"Indeks til minimum: {numpy_array.argmin()}")
print(f"Indeks til maximum: {numpy_array.argmax()}")
```

# Del 7: Hvorfor er NumPy raskere? (Eksempel)

```python
import numpy as np
import time

# Lag store lister/arrays
size = 1000000
python_list = list(range(size))
numpy_array = np.array(range(size))

# Test hastighet: Python list
start = time.time()
result_list = [x * 2 for x in python_list]
python_time = time.time() - start

# Test hastighet: NumPy array
start = time.time()
result_numpy = numpy_array * 2
numpy_time = time.time() - start

print(f"Python list tid: {python_time:.4f} sekunder")
print(f"NumPy array tid: {numpy_time:.4f} sekunder")
print(f"NumPy er {python_time/numpy_time:.1f}x raskere! 🚀")




```

## 📝 Key Takeaways

📝 Key Takeaways
Python Modules:

Module = En fil med Python-kode som kan importeres
Built-in modules: math, random, statistics (følger med Python)
User-defined modules: NumPy, Pandas, Matplotlib (må installeres med pip)
Import metoder:

import math → må bruke math.sqrt()
import math as m → kan bruke m.sqrt()
from math import sqrt → kan bruke sqrt() direkte


 Algebra Konsepter

Scalar: Et enkelt tall (0D) → 5
Vector: En liste med tall (1D) → [1, 2, 3, 4, 5]
Matrix: En tabell med tall (2D) → [[1,2,3], [4,5,6]]
Tensor: 3D eller høyere dimensjoner → [[[1,2],[3,4]], [[5,6],[7,8]]]

Data Science Packages:
NumPy (Numerical Python):

Hva: Arrays og matematikk - fundamentet for Data Science
Hvorfor: 10-100x raskere enn Python lists
Viktigste features:

Vectorized operations (operasjoner på hele arrays samtidig)
Memory efficient (bruker mindre minne)
Grunnlag for andre pakker (Pandas, Scikit-Learn bygger på NumPy)


Import: import numpy as np

Pandas (Panel Data):

Hva: Tabulære data og DataFrames (som Excel i Python)
Hvorfor: Lett å jobbe med tabeller, CSV, Excel-filer
Viktigste features:

DataFrames (tabeller med rader og kolonner)
Data manipulation (filtrering, sortering, gruppering)
Håndtere manglende data
Lese/skrive CSV, Excel, SQL


Import: import pandas as pd

SciPy (Scientific Python):

Hva: Avanserte algoritmer for vitenskap og ingeniørfag
Viktigste features:

Lineær algebra
Statistikk
Optimalisering
Integrasjon
Differensialligninger


Import: import scipy
Bygger på: NumPy

Scikit-Learn:

Hva: Machine Learning bibliotek
Viktigste features:

Klassifisering (f.eks. spam-deteksjon)
Regresjon (prediksjoner)
Clustering (gruppering av data)
Model validation


Import: from sklearn import ...
Bygger på: NumPy, SciPy, Matplotlib

Matplotlib:

Hva: 2D plotting bibliotek
Viktigste features:

Line plots
Scatter plots
Bar charts
Histograms
Pie charts


Import: import matplotlib.pyplot as plt
Lavnivå: Krever litt mer kode, men veldig fleksibelt

Seaborn:

Hva: Statistisk visualisering (bygger på Matplotlib)
Viktigste features:

Penere grafer "out of the box"
Høynivå interface
Spesialisert på statistiske plots


Import: import seaborn as sns
Bygger på: Matplotlib

Keras:

Hva: Neural networks og deep learning
Viktigste features:

Bygge nevrale nettverk enkelt
User-friendly API
Støtter TensorFlow, Theano, CNTK backends


Import: import keras eller from tensorflow import keras
