## 🧠 Concept Explanation (in my own words)

Describe the topic in your own words:

 What is this about?

 - This lecture introduces dictionaries, wich store data as key-value paris instead of indexed positions.

 Why is it important?

 - Dictionaries allow fast lookups by key and are ideal for representing structured data such as profiles, settings, or mappings.


 When or where is it used?
 
 - Dictionaries are used whenever data should be accessed by meaningful label (key) rather tan by position.


```python
# Code example / demonstration

# Creating a dictionary

student = {
    "name": "Shaffi",
    "age": 24,
    "course": "python"

}

print(student)
print(type(student))

# Accessing values
print(student["name"])
print(student.get("age"))

# Modifying values (dictionaries are mutable)
student["age"] = 25
print(student)

# Adding new key-value pair
student["grade"] = "A"
print(student)

# Removing a key-value pair
student.pop("course")
print(student)
```

## 📝 Key Takeaways

- Dictionaries are unordered (no fixed index order)
- Acess values using keys, not indices
- Dictionaries are mutable
- Keys must be unique and immutable (e.g., string, numbers) 


📝 Summary (Key Takeaways)

Dictionaries store data as key–value pairs

Values are accessed using keys, not indexes

Dictionaries are mutable and can be modified

Keys must be unique and immutable

Dictionaries are ideal for structured, labeled data
