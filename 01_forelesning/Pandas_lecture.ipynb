🧠 Concept Explanation (in my own words)

What is this about?
Pandas er Python-biblioteket for å jobbe med tabulære data (data i tabellform - som Excel!). Navnet kommer fra "Panel Data". Med Pandas kan du lese, analysere, filtrere, og manipulere data på en enkel og kraftig måte.


Why is it important?
I den virkelige verden kommer data ofte i tabeller - CSV-filer, Excel, databaser, etc. Pandas gjør det enkelt å:

Lese og skrive filer
Rense "skitne" data
Analysere store datasett
Lage rapporter og visualiseringer
Forberede data for machine learning

When or where is it used?

Data analyse: Analysere salgsdata, kundedata, etc.
Data science: Utforske datasett før modeling
Business intelligence: Lage rapporter fra rådata
Web scraping: Organisere data fra nettet
Forskning: Analysere eksperimentdata
